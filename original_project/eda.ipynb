```python
import re

# 1. Standardize judgment date
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# 2. Handle multiple hearing dates (if present in list form)
def get_earliest_hearing_date(h):
    if pd.isna(h) or h is None:
        return None

    # Convert to string
    h = str(h)

    # Split on commas, semicolons, or spaces
    parts = re.split(r'[;,]|  ', h)

    # Convert each to a date if possible
    valid_dates = []
    for p in parts:
        try:
            d = pd.to_datetime(p.strip(), errors='coerce')
            if not pd.isna(d):
                valid_dates.append(d)
        except:
            pass

    # Return earliest date or None
    return min(valid_dates) if valid_dates else None


# Apply the function
df['hearing_date_clean'] = df['hearing_date'].apply(get_earliest_hearing_date)

# 3. Calculate hearing gap safely
df['hearing_gap_days'] = (df['date'] - df['hearing_date_clean']).dt.days

# 4. Fix negative, impossible, or invalid values
df.loc[df['hearing_gap_days'] < 0, 'hearing_gap_days'] = np.nan

# 5. Final cleaning: If hearing_date missing, set hearing_gap to NaN
df.loc[df['hearing_date_clean'].isna(), 'hearing_gap_days'] = np.nan
```

```python
#  Legal Category Bucket (Top Section)
# -------------------------------------------------
def map_section_category(section):
    s = str(section).lower()
    if any(x in s for x in ['302', '376', '307', '498', '420', 'criminal']):
        return "Criminal"
    if any(x in s for x in ['civil', 'contract', 'property', 'family']):
        return "Civil"
    if any(x in s for x in ['article 14', 'article 21', 'constitution']):
        return "Constitutional"
    if any(x in s for x in ['administrative', 'service', 'office']):
        return "Administrative"
    return "Miscellaneous"

df['legal_category'] = df['top_section'].apply(map_section_category)

# Chart: Legal category distribution
plt.figure(figsize=(7,5))
sns.countplot(y=df['legal_category'])
plt.title("Distribution of Legal Categories")
plt.xlabel("Count")
plt.ylabel("Legal Category")
plt.show()
```

```python
def map_pakistan_section_category(section_text):
    if pd.isna(section_text):
        return "Miscellaneous"

    # extract integers
    nums = []
    for part in str(section_text).split(','):
        part = part.strip()
        if part.isdigit():
            nums.append(int(part))

    # 1. Constitutional Articles
    if 14 in nums or 21 in nums:
        return "Constitutional"

    # 2. Pakistan Penal Code (PPC) – Criminal
    if any(1 <= n <= 511 for n in nums):
        return "Criminal"

    # 3. Criminal Procedure Code (CrPC) – Criminal
    if any(1 <= n <= 565 for n in nums):
        return "Criminal"

    # 4. Civil Laws (Contract Act, Evidence Act, Family laws)
    if any(100 <= n <= 2000 for n in nums):
        return "Civil"

    # 5. Everything else → Special Acts / Misc
    if any(n > 2000 for n in nums):
        return "Miscellaneous"

    return "Miscellaneous"


df['legal_category'] = df['sections'].apply(map_pakistan_section_category)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight

from xgboost import XGBClassifier

# ---------------------------
# LOAD DATA
# ---------------------------
df = pd.read_csv("cleaned_out.csv")

# ---------------------------
# FIX MISSING TEXT VALUES
# ---------------------------
df['petition_text'] = df['petition_text'].fillna("")
df['sections'] = df['sections'].fillna("")

# ---------------------------
# TARGET VARIABLE
# ---------------------------
df = df[df['outcome'].notna()]
y = df['outcome']

# Encode labels
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# ---------------------------
# TEXT FEATURES (TF-IDF)
# ---------------------------
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2),
    stop_words='english'
)

X_text = tfidf.fit_transform(df['petition_text'])

# ---------------------------
# NUMERICAL FEATURES
# ---------------------------
numeric_cols = ['word_count', 'char_count', 'year', 'month', 'case_duration_days']
X_num = df[numeric_cols].fillna(0)

# Combine text + numeric
from scipy.sparse import hstack
X = hstack([X_text, X_num])

# ---------------------------
# TRAIN/TEST SPLIT
# ---------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# ---------------------------
# CLASS WEIGHT BALANCING
# ---------------------------
classes = np.unique(y_encoded)
weights = compute_class_weight("balanced", classes=classes, y=y_encoded)
class_weights = dict(zip(classes, weights))

print("Class weights:", class_weights)

# ---------------------------
# XGBOOST MODEL
# ---------------------------
model = XGBClassifier(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective='multi:softprob',
    num_class=len(classes),
    eval_metric='mlogloss',
    scale_pos_weight=1  # we manually apply weights below
)

# Fit with sample weights
sample_weights = np.array([class_weights[i] for i in y_train])
model.fit(X_train, y_train, sample_weight=sample_weights)

# ---------------------------
# EVALUATE
# ---------------------------
y_pred = model.predict(X_test)

print("\n=== CLASSIFICATION REPORT ===")
print(classification_report(y_test, y_pred, target_names=le.classes_))

print("\n=== CONFUSION MATRIX ===")
print(confusion_matrix(y_test, y_pred))
```
